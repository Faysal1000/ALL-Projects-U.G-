```python
import os
import shutil
from tqdm import tqdm

# -------- paths --------
SOURCE_DIR = r"archive\trainx"
MEL_LIST_FILE = "ph2 mel.txt"
MEL_DIR = "PH2-dataset/Malignant"
BENIGN_DIR = "PH2-dataset/Benign"
# ----------------------

# create output folders if not exist
os.makedirs(MEL_DIR, exist_ok=True)
os.makedirs(BENIGN_DIR, exist_ok=True)

# read melanoma image names
with open(MEL_LIST_FILE, "r") as f:
    mel_ids = set(line.strip() for line in f if line.strip())

print(f"Total melanoma IDs: {len(mel_ids)}")

# list images
image_files = [
    f for f in os.listdir(SOURCE_DIR)
    if f.lower().endswith(".bmp")
]

# move images with progress bar
for filename in tqdm(image_files, desc="Separating images", unit="img"):
    image_id = os.path.splitext(filename)[0]
    src_path = os.path.join(SOURCE_DIR, filename)

    if image_id in mel_ids:
        dst_path = os.path.join(MEL_DIR, filename)
    else:
        dst_path = os.path.join(BENIGN_DIR, filename)

    shutil.move(src_path, dst_path)

print("✅ Files successfully separated into mel / benign folders")
```
