```python
import os

def delete_superpixel_files(folder_path):
    """
    Deletes all files in the given folder (and its subfolders)
    that have 'superpixel' in their filename.
    """
    deleted_files = []

    for root, dirs, files in os.walk(folder_path):
        for file in files:
            if "superpixel" in file:
                file_path = os.path.join(root, file)
                try:
                    os.remove(file_path)
                    deleted_files.append(file_path)
                    print(f"Deleted: {file_path}")
                except Exception as e:
                    print(f"Error deleting {file_path}: {e}")

    if not deleted_files:
        print("No files containing 'superpixel' found.")
    else:
        print(f"\nTotal deleted files: {len(deleted_files)}")
```

```python
# -------------------------------
# usage
# -------------------------------
folder = r"ISIC_2017_Image_and_Mask\Images"  
delete_superpixel_files(folder)
```
