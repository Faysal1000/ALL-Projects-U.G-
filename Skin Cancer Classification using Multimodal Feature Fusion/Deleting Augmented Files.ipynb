```python
import os

def delete_aug_files_in_subfolders(main_folder):
    if not os.path.isdir(main_folder):
        print("Invalid folder path.")
        return

    # Loop through each item in the main folder
    for subfolder_name in os.listdir(main_folder):
        subfolder_path = os.path.join(main_folder, subfolder_name)

        # Check if it's a directory
        if os.path.isdir(subfolder_path):
            # List files inside the subfolder
            for file_name in os.listdir(subfolder_path):
                if file_name.startswith("aug"):
                    file_path = os.path.join(subfolder_path, file_name)
                    try:
                        os.remove(file_path)
                        #print(f"Deleted: {file_path}")
                    except Exception as e:
                        print(f"Failed to delete {file_path}: {e}")

# === Usage ===
delete_aug_files_in_subfolders(r"C:\Users\Faysal Ahmmed\Desktop\HAM10000_organized")
```
